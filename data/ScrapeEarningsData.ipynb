# Scrape Earnings Data using `yfinance`

```python
import pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed
import yfinance as yf
```

```python
df_r3000 = pd.read_csv("iShares-Russell-3000-ETF_fund.csv")
tickers = df_r3000['Ticker'].tolist()
print(len(tickers))
```

Split the tickers into batches of ~400 to avoid rate limits from yfinance

```python
tickers1 = tickers[:400]
tickers2 = tickers[400:800]
tickers3 = tickers[800:1200]
tickers4 = tickers[1200:1600]
tickers5 = tickers[1600:2000]
tickers6 = tickers[2000:2571]
```

```python
def fetch_ticker_data(t):
    try:
        ticker_obj = yf.Ticker(t)
        df = ticker_obj.get_earnings_dates(limit=4)
        date = df.index[df.index.year == 2026][-1]
        eps_estimate = df.loc[date, 'EPS Estimate']
        eps_reported = df.loc[date, 'Reported EPS']
        surprise_pct = ((eps_reported - eps_estimate) / abs(eps_estimate)) * 100 if eps_estimate != 0 else None
        return t, (date, eps_estimate, eps_reported, surprise_pct)
    except Exception as e:
        print(f"Error fetching data for {t}: {e}")
        return t, None
```

## Get earnings data in batches

```python
t1_earnings = {}
with ThreadPoolExecutor(max_workers=8) as executor:
    futures = {executor.submit(fetch_ticker_data, t): t for t in tickers1}
    for future in as_completed(futures):
        t, result = future.result()
        if result is not None:
            t1_earnings[t] = result
```

```python
df_t1 = pd.DataFrame.from_dict(t1_earnings, orient='index', columns=['Date', 'EPS Estimate', 'EPS Reported', 'Surprise%']).reset_index().rename(columns={'index': 'Ticker'})
print(df_t1.head())
```

```python
df_t1.to_csv("earnings1.csv", index=False)
```

Get remaining tickers for which data wasn't scraped

```python
earnings1_tickers = set(pd.read_csv("earnings1.csv")['Ticker'])
earnings2_tickers = set(pd.read_csv("earnings2.csv")['Ticker'])
earnings3_tickers = set(pd.read_csv("earnings3.csv")['Ticker'])
earnings4_tickers = set(pd.read_csv("earnings4.csv")['Ticker'])
earnings5_tickers = set(pd.read_csv("earnings5.csv")['Ticker'])
earnings6_tickers = set(pd.read_csv("earnings6.csv")['Ticker'])

tickers_remaining = set(tickers) - earnings1_tickers - earnings2_tickers - earnings3_tickers - earnings4_tickers - earnings5_tickers - earnings6_tickers
```

```python
print(f"Tickers remaining: {len(tickers_remaining)}")
```

Try to fetch data again (in case they weren't scraped due to rate limits) and track tickers for which there is no data

```python
trem_earnings = {}
excluded_tickers = []

with ThreadPoolExecutor(max_workers=8) as executor:
    futures = {executor.submit(fetch_ticker_data, t): t for t in tickers_remaining}
    for future in as_completed(futures):
        t, result = future.result()
        if result is not None:
            trem_earnings[t] = result
        else:
            excluded_tickers.append(t)
```

```python
df_trem_earnings = pd.DataFrame.from_dict(trem_earnings, orient='index', columns=['Date', 'EPS Estimate', 'EPS Reported', 'Surprise%']).reset_index().rename(columns={'index': 'Ticker'})
df_trem_earnings.to_csv("earnings7.csv", index=False)
```

```python
import glob

# Read all earnings CSV files
earnings_files = glob.glob('earnings*.csv')
dfs = [pd.read_csv(file) for file in earnings_files]

# Merge all dataframes
merged_df = pd.concat(dfs, ignore_index=True)

# Save to a new CSV file
merged_df.to_csv('earnings_merged.csv', index=False)

print(f"Merged {len(dfs)} files into earnings_merged.csv")
print(f"Total rows: {len(merged_df)}")
```
