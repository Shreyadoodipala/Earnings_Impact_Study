# Clean Earnings Data

```python
import pandas as pd
import numpy as np
```

```python
df_earnings = pd.read_csv('earnings_merged.csv')
df_earnings.info()
```

```python
df_earnings['Date'] = pd.to_datetime(df_earnings['Date'], utc=True).dt.strftime('%Y-%m-%d')
```

```python
df_earnings['Surprise%'] = np.where(df_earnings['EPS Estimate'] != 0,
    ((df_earnings['EPS Reported'] - df_earnings['EPS Estimate']) / abs(df_earnings['EPS Estimate'])) * 100, None)
```

```python
# display all rows
pd.set_option('display.max_rows', None)
```

```python
print(df_earnings.isnull().sum())
print("Total rows with missing values:", df_earnings.isnull().any(axis=1).sum())
df_earnings[df_earnings.isnull().any(axis=1)]
```

```python
# Drop rows with missing values and round 'Surprise%' to 4 decimal places
df_earnings_modified = df_earnings.dropna()
df_earnings_modified['Surprise%'] = df_earnings_modified['Surprise%'].round(4)

print("Total rows after dropping missing values:", df_earnings_modified.shape[0])
df_earnings_modified.head()
```

```python
df_earnings_modified.to_csv('earnings_cleaned.csv', index=False)
```

```python
# Load the Russell 3000 ETF fund data and filter it to include only tickers present in the cleaned earnings data
df_russell_shares = pd.read_csv('iShares-Russell-3000-ETF_fund.csv')
df_russell_shares = df_russell_shares[df_russell_shares['Ticker'].isin(df_earnings_modified['Ticker'])]
df_russell_shares.to_csv('iShares-Russell-3000-ETF_fund_cleaned.csv', index=False)
```
